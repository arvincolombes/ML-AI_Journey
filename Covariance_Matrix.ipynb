## Covariance Matrix
When we have multiple variables, we compute covariance between every pair.
The result is a square matrix called the covariance matrix.
A covariance matrix is a square matrix that shows how different variables in a dataset vary together



  
Where X1 and Yy are elements in data

```python
from sklearn.datasets import fetch_openml
mnist = fetch_openml("mnist_784", version=1, as_frame=False)
```

```python
mnist.data, mnist.target
```

```python
mnist.data[0].shape
```

```python
mnist.target[0]
```

```python
x = mnist.data
y = mnist.target
```

```python
x.shape, x[0].shape, y.shape
```

```python
import matplotlib.pyplot as plt
fig, axes = plt.subplots(2, 5, figsize=(10, 4))
for i, ax in enumerate(axes.ravel()):
    ax.imshow(x[i].reshape(28,28),cmap="gray")
    ax.set_title(y[i])
    ax.axis("off")
plt.tight_layout()
plt.show()
```
